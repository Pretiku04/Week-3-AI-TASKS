```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("Student_Performance.csv")

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

```python
print("Dataset Information:")
print(df.info())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
features = [
    "Study_Hours",
    "Attendance",
    "Previous_Marks",
    "Assignment_Score"
]

X = df[features]
y = df["Result"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Encoded target values:")
print(y_encoded)

print("\nClasses:")
print(label_encoder.classes_)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
model = DecisionTreeClassifier(
    random_state=42,
    max_depth=3
)

print("Decision Tree Classifier created successfully!")
```

```python
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("Predicted values:")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy (%):", accuracy * 100)
```

```python
new_students = pd.DataFrame({
    "Study_Hours": [6, 2, 8],
    "Attendance": [85, 60, 92],
    "Previous_Marks": [72, 45, 88],
    "Assignment_Score": [75, 50, 90]
})

predictions = model.predict(new_students)

new_students["Prediction"] = label_encoder.inverse_transform(predictions)

print("Predictions for New Students:")
print(new_students)
```

```python
student = pd.DataFrame({
    "Study_Hours": [6],
    "Attendance": [85],
    "Previous_Marks": [72],
    "Assignment_Score": [75]
})

prediction = model.predict(student)

result = label_encoder.inverse_transform(prediction)[0]

print("New Student Details:")
print(student)

print("\nPrediction:", result.upper())
```
