```python
# Create a copy so the original dataset remains unchanged
df_missing = df.copy()

# Create some missing values for demonstration
df_missing.loc[2, "SepalLengthCm"] = None
df_missing.loc[10, "SepalWidthCm"] = None
df_missing.loc[20, "PetalLengthCm"] = None
df_missing.loc[30, "PetalWidthCm"] = None

print("Missing values in each column:")
print(df_missing.isnull().sum())
```

```python
# Fill missing numerical values with the mean of their respective columns

numerical_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

for column in numerical_columns:
    df_missing[column] = df_missing[column].fillna(df_missing[column].mean())

print("Missing values after cleaning:")
print(df_missing.isnull().sum())
```

```python
features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

target = "Species"

print("Features:")
print(features)

print("\nTarget:")
print(target)
```
